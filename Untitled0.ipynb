```python
import random
from enum import IntEnum
UserScore=0
ComputerScore=0
class Action(IntEnum):
    Rock = 0
    Paper = 1
    Scissors = 2
def get_user_selection():
    choices = [f"{action.name}[{action.value}]" for action in Action]
    choices_str = ", ".join(choices)
    selection = int(input(f"Enter a choice: ({choices_str}): "))
    action = Action(selection)
    return action
def get_computer_selection():
    selection = random.randint(0, len(Action) - 1)
    action = Action(selection)
    return action
def determine_winner(user_action, computer_action):
    global UserScore, ComputerScore
    if user_action == computer_action:
        print(f"Both players selected {user_action.name}. It's a tie!")
    elif user_action == Action.Rock:
        if computer_action == Action.Scissors:
            print("Rock smashes scissors! One point for you!")
            UserScore=UserScore+1
        else:
            print("Paper covers rock! One point for the computer.")
            ComputerScore=ComputerScore+1
    elif user_action == Action.Paper:
        if computer_action == Action.Rock:
            print("Paper covers rock! One point for you!")
            UserScore=UserScore+1
        else:
            print("Scissors cuts paper! One point for the computer.")
            ComputerScore=ComputerScore+1
    elif user_action == Action.Scissors:
        if computer_action == Action.Paper:
            print("Scissors cuts paper! One point for you!")
            UserScore=UserScore+1
        else:
            print("Rock smashes scissors! One point for the computer.")
            ComputerScore=ComputerScore+1

while True:
    UserScore = 0
    ComputerScore = 0
    print("Welcome to Rock Paper Scissors!")
    print("Your goal is to beat the computer, best two out of three.")

    while(1):
        try:
            user_action = get_user_selection()
        except ValueError as e:
            range_str = f"[0, {len(Action) - 1}]"
            print(f"Invalid selection. Enter a value in range {range_str}")
            continue

        computer_action = get_computer_selection()
        determine_winner(user_action, computer_action)
        print(f"User: {UserScore} points. Computer: {ComputerScore} points.\n")

        if UserScore == 2:
            print("\nCongratulations! You reached 2 points and won the game!")
            break
        elif ComputerScore == 2:
            print("\nGame over! The computer reached 2 points and won the game.")
            break

    while True:
        play_again = input("Would you like to play another match? (y/n): ").strip().lower()
        if play_again in ['y', 'n']:
            break
        print("Invalid input. Please enter 'y' for yes or 'n' for no.")

    if play_again == 'n':
        print("Thanks for playing!")
        break
```
